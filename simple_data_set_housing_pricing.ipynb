```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df= pd.read_csv("/content/drive/MyDrive/Car Sales Price Prediction/house_price.csv")
df
```

```python
plt.xlabel("price", color = "red")
plt.ylabel("area", color = "red")
plt.scatter(df.price,df.area)
```

```python
lr= linear_model.LinearRegression()
lr.fit(df[['area']],df.price)
```

```python
lr.score(df[['area']],df.price)
```

```python
predidicted= lr.predict(df[['area']])
print(predidicted)
```

```python
plt.plot(df.area,lr.predict(df[['area']]),color="blue")
plt.xlabel("price", color = "red")
plt.ylabel("area", color = "red")
plt.scatter(df.area,df.price)
```
